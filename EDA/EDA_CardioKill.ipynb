Import Required Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

%matplotlib inline
```

Load the Dataset

```python
df = pd.read_csv("/content/drive/MyDrive/Colab Notebooks/DSGP/Heart_health new.csv")
```

Dataset Overview

```python
df.head()

#Dataset Shape
df.shape
```

Data Types & Structure

```python
df.info()
```

Statistical Summary

```python
df.describe()
```

Missing Value Analysis

```python
df.isnull().sum()
```

Target Variable Analysis (Heart Attack)

```python
df['Heart Attack'].value_counts()
df['Heart Attack'].value_counts(normalize=True)
sns.countplot(x='Heart Attack', data=df)
plt.title("Heart Attack Class Distribution")
plt.show()
```

Univariate Analysis (Numerical Features)

```python
#Age Distribution
plt.hist(df['Age'], bins=20)
plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()

#Cholesterol Distribution
plt.hist(df['Cholesterol mg/dL'], bins=20)
plt.title("Cholesterol Distribution")
plt.xlabel("Cholesterol (mg/dL)")
plt.ylabel("Frequency")
plt.show()

#Exercise Hours per Week
plt.hist(df['Exercise hours/week'], bins=20)
plt.title("Exercise Hours per Week")
plt.xlabel("Hours")
plt.ylabel("Frequency")
plt.show()

```

Categorical Feature Analysis

```python
#Gender Distribution
df['Gender'].value_counts()

sns.countplot(x='Gender', data=df)
plt.title("Gender Distribution")
plt.show()


#Smoking Status
df['Smoker'].value_counts()

sns.countplot(x='Smoker', data=df)
plt.title("Smoking Status Distribution")
plt.show()
```

Bivariate Analysis (Feature vs Target)

```python
#Age vs Heart Attack
sns.boxplot(x='Heart Attack', y='Age', data=df)
plt.title("Age vs Heart Attack")
plt.show()

#Cholesterol vs Heart Attack
sns.boxplot(x='Heart Attack', y='Cholesterol mg/dL', data=df)
plt.title("Cholesterol vs Heart Attack")
plt.show()

#Exercise vs Heart Attack
sns.boxplot(x='Heart Attack', y='Exercise hours/week', data=df)
plt.title("Exercise Hours vs Heart Attack")
plt.show()
```

Blood Pressure Feature Analysis

```python
#Split Blood Pressure into Systolic & Diastolic
df[['Systolic_BP', 'Diastolic_BP']] = df['Blood Pressure mmHg'] \
    .str.split('/', expand=True).astype(int)

df[['Systolic_BP', 'Diastolic_BP']].describe()
```

Correlation Analysis

```python
plt.figure(figsize=(10,6))
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap='coolwarm')
plt.title("Correlation Heatmap")
plt.show()
```
